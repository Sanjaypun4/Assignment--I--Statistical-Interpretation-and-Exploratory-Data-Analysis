Problem 2 — Advanced HDI Exploration

Task 1: Create South Asia Subset

```python
south_asia_countries = ["Afghanistan", "Bangladesh", "Bhutan", "India",
                        "Maldives", "Nepal", "Pakistan", "Sri Lanka"]

# Filter dataset for South Asia
hdi_south_asia = df[df['country'].isin(south_asia_countries)]

hdi_south_asia = hdi_south_asia.drop_duplicates()

# Save as CSV
hdi_south_asia.to_csv("HDI_SouthAsia.csv", index=False)

# Preview
hdi_south_asia.head()
```

Task 2: Composite Development Score

```python
hdi_south_asia['life_expectancy'] = pd.to_numeric(hdi_south_asia['life_expectancy'], errors='coerce')
hdi_south_asia['gross_inc_percap'] = pd.to_numeric(hdi_south_asia['gross_inc_percap'], errors='coerce')

# Drop rows with missing values in required columns
hdi_south_asia = hdi_south_asia.dropna(subset=['life_expectancy', 'gross_inc_percap'])

# Calculate Composite Score
hdi_south_asia['Composite Score'] = (0.30 * hdi_south_asia['life_expectancy'] +
                                     0.30 * hdi_south_asia['gross_inc_percap'])

# Rank countries by Composite Score
hdi_south_asia = hdi_south_asia.sort_values(by='Composite Score', ascending=False)

# Horizontal bar chart of top 5 countries
plt.figure(figsize=(8,5))
sns.barplot(data=hdi_south_asia.head(5), x='Composite Score', y='country', palette='viridis')
plt.title("Task 2: Top 5 South Asian Countries by Composite Score")
plt.xlabel("Composite Score")
plt.ylabel("Country")
plt.show()

# Preview table
hdi_south_asia[['country', 'Composite Score', 'hdi']].head(5)
```

Problem 2 — Task 3: Outlier Detection

```python

def detect_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series[(series < lower_bound) | (series > upper_bound)]

# Detect outliers in HDI
hdi_outliers = detect_outliers(hdi_south_asia['hdi'])
print("HDI Outliers:")
print(hdi_outliers)

# Detect outliers in GNI per Capita
gni_outliers = detect_outliers(hdi_south_asia['gross_inc_percap'])
print("\nGNI per Capita Outliers:")
print(gni_outliers)

# Scatter plot: GNI per Capita vs HDI, highlighting outliers
plt.figure(figsize=(8,5))
sns.scatterplot(data=hdi_south_asia, x='gross_inc_percap', y='hdi', label='Normal', color='blue')

# Highlight HDI or GNI outliers in red
outlier_data = hdi_south_asia[hdi_south_asia['country'].isin(
    list(hdi_outliers.index) + list(gni_outliers.index)
)]
sns.scatterplot(data=outlier_data, x='gross_inc_percap', y='hdi', label='Outlier', color='red')

plt.xlabel("GNI per Capita")
plt.ylabel("HDI")
plt.title("Task 3: HDI vs GNI per Capita with Outliers Highlighted")
plt.legend()
plt.show()
```

Problem 2 — Task 4: Exploring Metric Relationships

```python

hdi_south_asia['gender_development'] = pd.to_numeric(hdi_south_asia['gender_development'], errors='coerce')
hdi_south_asia['life_expectancy'] = pd.to_numeric(hdi_south_asia['life_expectancy'], errors='coerce')
hdi_south_asia = hdi_south_asia.dropna(subset=['gender_development', 'life_expectancy', 'hdi'])

# Pearson correlation with HDI
corr_gender = hdi_south_asia['gender_development'].corr(hdi_south_asia['hdi'])
corr_life = hdi_south_asia['life_expectancy'].corr(hdi_south_asia['hdi'])

print(f"Correlation (HDI vs Gender Development Index): {corr_gender:.2f}")
print(f"Correlation (HDI vs Life Expectancy): {corr_life:.2f}")

# Scatter plot: HDI vs Gender Development Index
plt.figure(figsize=(7,5))
sns.scatterplot(data=hdi_south_asia, x='gender_development', y='hdi')
sns.regplot(data=hdi_south_asia, x='gender_development', y='hdi', scatter=False, color='red')
plt.xlabel("Gender Development Index")
plt.ylabel("HDI")
plt.title("Task 4a: HDI vs Gender Development Index")
plt.show()

# Scatter plot: HDI vs Life Expectancy
plt.figure(figsize=(7,5))
sns.scatterplot(data=hdi_south_asia, x='life_expectancy', y='hdi')
sns.regplot(data=hdi_south_asia, x='life_expectancy', y='hdi', scatter=False, color='red')
plt.xlabel("Life Expectancy")
plt.ylabel("HDI")
plt.title("Task 4b: HDI vs Life Expectancy")
plt.show()
```

Problem 2 — Task 5: Gap Analysis

```python
# Ensure numeric GNI and HDI
hdi_south_asia['gross_inc_percap'] = pd.to_numeric(hdi_south_asia['gross_inc_percap'], errors='coerce')
hdi_south_asia = hdi_south_asia.dropna(subset=['gross_inc_percap', 'hdi'])

# Create GNI HDI Gap
hdi_south_asia['GNI_HDI_Gap'] = hdi_south_asia['gross_inc_percap'] - hdi_south_asia['hdi']

# Rank countries by GNI HDI Gap
gap_sorted = hdi_south_asia.sort_values(by='GNI_HDI_Gap', ascending=False)
print("Top 3 Positive GNI HDI Gaps:")
print(gap_sorted[['country', 'GNI_HDI_Gap']].head(3))

print("\nTop 3 Negative GNI HDI Gaps:")
print(gap_sorted[['country', 'GNI_HDI_Gap']].tail(3))

# Plot Top 3 Positive Gaps
plt.figure(figsize=(8,4))
sns.barplot(
    data=gap_sorted.head(3),
    x='country', y='GNI_HDI_Gap', palette='Greens'
)
plt.title("Top 3 Positive GNI HDI Gaps in South Asia")
plt.ylabel("GNI - HDI Gap")
plt.show()

# Plot Top 3 Negative Gaps
plt.figure(figsize=(8,4))
sns.barplot(
    data=gap_sorted.tail(3),
    x='country', y='GNI_HDI_Gap', palette='Reds'
)
plt.title("Top 3 Negative GNI HDI Gaps in South Asia")
plt.ylabel("GNI - HDI Gap")
plt.show()
```

Problem 3 — Task 6: Correlation Analysis

```python

# Function to compute correlation and plot scatter
def correlation_plot(region_df, region_name):
    metrics = ['gender_development', 'life_expectancy']
    for metric in metrics:
        corr = region_df['hdi'].corr(region_df[metric])
        plt.figure(figsize=(6,4))
        sns.scatterplot(data=region_df, x=metric, y='hdi')
        sns.regplot(data=region_df, x=metric, y='hdi', scatter=False, color='red')
        plt.title(f"{region_name}: HDI vs {metric} (r = {corr:.2f})")
        plt.xlabel(metric)
        plt.ylabel("HDI")
        plt.show()
        print(f"{region_name} — Pearson correlation between HDI and {metric}: {corr:.2f}")

# South Asia correlation plots
correlation_plot(hdi_south_asia_2020_2022, "South Asia")

# Middle East correlation plots
correlation_plot(hdi_middle_east_2020_2022, "Middle East")
```

Problem 3 — Task 7: Outlier Detection

```python


def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

# Detect outliers for South Asia
outliers_hdi_sa = detect_outliers(hdi_south_asia_2020_2022, 'hdi')
outliers_gni_sa = detect_outliers(hdi_south_asia_2020_2022, 'gross_inc_percap')

# Detect outliers for Middle East
outliers_hdi_me = detect_outliers(hdi_middle_east_2020_2022, 'hdi')
outliers_gni_me = detect_outliers(hdi_middle_east_2020_2022, 'gross_inc_percap')

# Scatter plot HDI vs GNI per Capita with outliers highlighted
def plot_outliers(df, region_name):
    plt.figure(figsize=(7,5))
    sns.scatterplot(data=df, x='gross_inc_percap', y='hdi', label='Normal', color='blue')

    # Highlight outliers
    hdi_out = detect_outliers(df, 'hdi')
    gni_out = detect_outliers(df, 'gross_inc_percap')

    sns.scatterplot(data=hdi_out, x='gross_inc_percap', y='hdi', label='HDI Outliers', color='red', s=100)
    sns.scatterplot(data=gni_out, x='gross_inc_percap', y='hdi', label='GNI Outliers', color='green', s=100)

    plt.title(f"{region_name}: HDI vs GNI per Capita with Outliers")
    plt.xlabel("GNI per Capita")
    plt.ylabel("HDI")
    plt.legend()
    plt.show()

# Plot for South Asia
plot_outliers(hdi_south_asia_2020_2022, "South Asia")

# Plot for Middle East
plot_outliers(hdi_middle_east_2020_2022, "Middle East")
```
